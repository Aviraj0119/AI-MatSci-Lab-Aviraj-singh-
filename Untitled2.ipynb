```python
import pandas as pd
```

```python
# 1. Define the dictionary with 5 material properties
```

```python
materials_dict = {
    'Element': ['Fe', 'Al', 'Ti'],
    'Atomic_Number': [26, 13, 22],
    'Density_g_cm3': [7.87, 2.70, 4.51],
    'Melting_Point_K': [1811, 933, 1941],
    'Crystal_Structure': ['BCC', 'FCC', 'HCP'],
    'Bandgap_eV': [0.0, 0.0, 0.0] # Metals have no bandgap
}
```

```python
# 2. Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(materials_dict) 
print("\n--- DataFrame Info ---")
print(df.info()) 

print("\n--- Data Types ---")
print(df.dtypes) 

print("\n--- DataFrame Shape ---")
print(df.shape) 

print("\n--- Statistical Description ---")
print(df.describe())
```
