```python
import pandas as pd
from data_profiling import ProfileReport
from sklearn.impute import SimpleImputer, KNNImputer
```

```python
# 1. Load the dataset from Session S09
df = pd.read_csv("perovskites_data.csv")
```

```python
# 2. Generate the Automated Profiling Report
profile = ProfileReport(df, title="Perovskites Data Profiling Report")
profile.to_file("perovskites_report.html")
print("✅ Profiling report generated and saved as HTML.")
```

```python
print(df.columns.tolist())
```

```python
# 3. Handle Missing Values: Mean Imputation
mean_imputer = SimpleImputer(strategy='mean')

df_mean_imputed = df.copy()

# Impute missing values in Band_Gap column
df_mean_imputed[['Band_Gap']] = mean_imputer.fit_transform(
    df[['Band_Gap']]
)

print("Mean imputation completed successfully!")
```

```python
# 4. Handle Missing Values: KNN Imputation
knn_imputer = KNNImputer(n_neighbors=5)

df_knn_imputed = df.copy()

df_knn_imputed[['Band_Gap']] = knn_imputer.fit_transform(
    df[['Band_Gap']]
)

print("Missing data successfully imputed using Mean and KNN strategies.")
```
